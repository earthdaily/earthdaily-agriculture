## Step 3: Run Workflow

```python
manager.load_workflow("configuration/workflow.yml")
manager.visualize_workflow().show()
```

```python
results = manager.run_workflow(entity_list=entities)
```

## Step 4: Explore Results

```python
for step_name, step_result in results.items():
    df = step_result.get('results_df', pd.DataFrame())
    errors = step_result.get('global_errors', [])
    print(f"  {step_name}: {len(df)} rows, {len(errors)} errors")
```
